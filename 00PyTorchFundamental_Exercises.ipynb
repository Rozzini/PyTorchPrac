```python
## 6. Create two random tensors of shape (2, 3) and send them both to the GPU 
##    (you'll need access to a GPU for this). Set torch.manual_seed(1234) 
##    when creating the tensors (this doesn't have to be the GPU random seed). The output should be something like:
##    Device: cuda
##    (tensor([[0.0290, 0.4019, 0.2598],
##           [0.3666, 0.0583, 0.7006]], device='cuda:0'),
##     tensor([[0.0518, 0.4681, 0.6738],
##           [0.3315, 0.7837, 0.5631]], device='cuda:0'))


# Set random seed
torch.cuda.manual_seed(1234)
torch.manual_seed(1234)

# Check for access to GPU
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Device: {device}")

# Create two random tensors on GPU
X = torch.rand(2,3).to(device)
Y = torch.rand(2,3).to(device)
X, Y
```

```python
## 7. Perform a matrix multiplication on the tensors you created in 6 (again, you may have to adjust the shapes of one of the tensors).
##     The output should look like:
##    (tensor([[0.3647, 0.4709],
##           [0.5184, 0.5617]], device='cuda:0'), torch.Size([2, 2]))

Z = torch.mm(X, Y.T)
Z, Z.shape
```

```python
## 8. Find the maximum and minimum values of the output of 7.

Zmax = Z.max()
Zmin = Z.min()
print("Max val: ", Zmax)
print("Min val: ", Zmin)
```

```python
## 9. Find the maximum and minimum index values of the output of 7.

Zargmax = Z.argmax()
Zargmin = Z.argmin()
print("Max pos: ", Zargmax)
print("Min pos: ", Zargmin)
```
