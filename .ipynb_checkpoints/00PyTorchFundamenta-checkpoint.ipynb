```python
## Creating ranged tensor

zero_to_ten = torch.arange(start=0, end=10, step=1)
zero_to_ten
```

```python
# Getting information from tensors

some_tensor = torch.rand(3,4)

print(some_tensor)
print(f"Shape: {some_tensor.shape}")
print(f"Datatype: {some_tensor.dtype}")
print(f"Device:  {some_tensor.device}")
```

```python
## add, minus, multiply
tensor = torch.tensor([2,5,6])
tensor_add = tensor + 5
tensor_minus = tensor - 5
tensor_mult = tensor * 5
tensor_div = tensor / 5
tensor_addf = torch.add(tensor, 5)
tensor_multf = torch.multiply(tensor, 5)
tensor_add, tensor_minus, tensor_mult, tensor_div, tensor_addf, tensor_multf
```

```python
## Matrix multiplication, most common operation
## torch.matmul()
tensor = torch.tensor([1, 2, 3])
tensor.shape
```

```python
# Element-wise matrix multiplication
tensor * tensor
```

```python
# Matrix multiplication
torch.matmul(tensor, tensor)
```

```python

### MAKE SURE SHAPES IS CORRECT

tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10],
                         [8, 11], 
                         [9, 12]], dtype=torch.float32)

torch.matmul(tensor_A, tensor_B) # (this will error)
```

```python

### ONE OF THE WAYS TO FIX IT IS TO TRANSPOSE ONE OF MATRICES, USING .T

tensor_A.shape, tensor_B.shape, tensor_B.T.shape

##after this operation we can do @ operaton
```

```python
output = torch.matmul(tensor_A, tensor_B.T)
print(output) 
```

```python
# torch.mm is a shortcut for matmul
torch.mm(tensor_A, tensor_B.T)
```

```python
tensor23 = torch.tensor([[1,2,3],
                        [4,5,6]])
tensor33 = torch.tensor([[3,6,9],
                        [1,4,7],
                        [2,5,8]])
torch.mm(tensor23, tensor33)
```

```python

### Finding the min, max, mean, sum, etc (aggregation)

```

```python
# Create a tensor
x = torch.arange(0, 100, 10)
x
```

```python
print(f"Minimum: {x.min()}")
print(f"Maximum: {x.max()}")
# print(f"Mean: {x.mean()}") # this will error
print(f"Mean: {x.type(torch.float32).mean()}") # won't work without float datatype
print(f"Sum: {x.sum()}")
```

```python
tensor = torch.arange(0,100,5)
print(f"Tensor: {tensor}")

# Returns index of max and min values
print(f"Index where max value occurs: {tensor.argmax()}")
print(f"Index where min value occurs: {tensor.argmin()}")
```

```python

## Change tensor datatype if tensors in different datatypes.

# Create a tensor and check its datatype
tensor = torch.arange(10., 100., 10.)
tensor.dtype
```
